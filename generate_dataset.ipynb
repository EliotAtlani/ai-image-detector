```python
import os
import shutil
import random
```

```python
def split_dataset(source_dir, train_dir, test_dir, train_ratio=0.8, seed=42):
    
    random.seed(seed)

    os.makedirs(train_dir, exist_ok=True)
    os.makedirs(test_dir, exist_ok=True)


    for category in os.listdir(source_dir):
        category_path = os.path.join(source_dir, category)
        if not os.path.isdir(category_path):
            continue 

        for subcat in os.listdir(category_path):
            subcat_path = os.path.join(category_path, subcat)
            if not os.path.isdir(subcat_path):
                continue  

            files = [f for f in os.listdir(subcat_path) if os.path.isfile(
                os.path.join(subcat_path, f))]
            if not files:
                print(f"No files in {subcat_path}. Skipping...")
                continue

            random.shuffle(files)
            split_idx = int(len(files) * train_ratio)
            train_files = files[:split_idx]
            test_files = files[split_idx:]

            train_subdir = os.path.join(train_dir, category, subcat)
            test_subdir = os.path.join(test_dir, category, subcat)
            os.makedirs(train_subdir, exist_ok=True)
            os.makedirs(test_subdir, exist_ok=True)

            for fname in train_files:
                src_path = os.path.join(subcat_path, fname)
                dst_path = os.path.join(train_subdir, fname)
                shutil.copy2(src_path, dst_path)

            for fname in test_files:
                src_path = os.path.join(subcat_path, fname)
                dst_path = os.path.join(test_subdir, fname)
                shutil.copy2(src_path, dst_path)

            print(
                f"Category '{category}/{subcat}': {len(train_files)} train, {len(test_files)} test files copied.")
```

```python
split_dataset("../data_resized", "../dataset_train", "../dataset_test", train_ratio=0.8)
```
